# Data Transformation
Data transformation involves modifying the structure, format, or values of data to make it more suitable for analysis and modeling. Properly transformed data can significantly improve the performance and convergence speed of machine learning algorithms.

In this session, we will focus on **Data Scaling**, which is one of the most common transformation techniques.

### Why Data Transformation?
- **Algorithm Compatibility**: Many algorithms (like SVM, KNN, and Neural Networks) are sensitive to the scale of features.
- **Improved Convergence**: Scaling helps gradient-descent-based algorithms reach the optimal solution faster.
- **Reduced Bias**: Ensures that features with larger ranges do not dominate those with smaller ranges.

```python
#  x = (x-u)/o 
import pandas as pd
from sklearn.preprocessing import StandardScaler,MinMaxScaler,MaxAbsScaler

df = {
    'age': [25,30,35,40,45],
    'height': [165,170,175,180,185],
    'weight': [70,75,80,85,90]
}

df = pd.DataFrame(df)
```

### 1. Standardization (Standard Scaling)
Standardization centers the data around a mean of 0 with a standard deviation of 1. It is most effective when the data follows a Gaussian (normal) distribution.

- **Formula**: $Z = \frac{(x - \mu)}{\sigma}$
- **Library**: `sklearn.preprocessing.StandardScaler`

```python
scaler = StandardScaler()
# fit the scaler
scaled_df = scaler.fit_transform(df)
scaled_df = pd.DataFrame(scaled_df, columns=df.columns)
scaled_df
```

### 2. Min-Max Scaling (Normalization)
Min-Max scaling transforms features by scaling them to a specific fixed range, typically **[0, 1]**. It is useful when the algorithm does not handle negative values or when you want to preserve the relative distances between values.

- **Formula**: $x_{scaled} = \frac{(x - x_{min})}{(x_{max} - x_{min})}$

```python
# MIN/MAX
scaler = MinMaxScaler()
# fit the scaler
scaled_df = scaler.fit_transform(df)
scaled_df = pd.DataFrame(scaled_df, columns=df.columns)
scaled_df
```

### 3. Max-Abs Scaling
Max-Abs scaling scales each feature by its maximum absolute value. This means the values will range between **[-1, 1]**. It is particularly useful for sparse data and does not shift or center the data, preserving the zero-points.

- **Formula**: $x_{scaled} = \frac{x}{|x_{max}|}$

```python
# MAX/ABS
scaler = MaxAbsScaler()
# fit the scaler
scaled_df = scaler.fit_transform(df)
scaled_df = pd.DataFrame(scaled_df, columns=df.columns)
scaled_df
```
