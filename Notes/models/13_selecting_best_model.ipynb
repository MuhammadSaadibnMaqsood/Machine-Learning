**SELECTING BEST MODEL**

To select the best model when using multiple model in pipline, you can use techniques such as cross-validation, grid search, or random search to evaluate the performance of each model. Here are some steps to follow:

```python
import pandas as pd
import numpy as np
import seaborn as sns
from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score
```

```python
df = sns.load_dataset('titanic')
```

```python
# split into x and y 

x = df[['pclass', 'age','sex', 'fare', 'embarked']]
y = df['survived']
```

```python
# train test split 

x_train, x_test, y_train,y_test = train_test_split(x,y,test_size=0.2, random_state=42)
```

```python
# models 

models = [
    ('Random forest', RandomForestClassifier(random_state=42)),
    ('Gradient Boosting', GradientBoostingClassifier(random_state=42))
]

best_model = None
best_accuracy = 0.0

for name, model in models:
    pipeline = Pipeline([
        ('impute', SimpleImputer(strategy='most_frequent')),
        ('encoder', OneHotEncoder(handle_unknown='ignore')),
        ('model', model)
    ])
    
    
    scores = cross_val_score(pipeline, x_train, y_train, cv=5)
    mean_score = scores.mean()

    pipeline.fit(x_train,y_train)

    pred = pipeline.predict(x_test)

    accuracy = accuracy_score(y_test, pred)

    print('Model Name: ', name)
    print('Cross Validation Accuracy: ',mean_score)
    print('Test Accuracy: ', accuracy)
    print()

    if accuracy > best_accuracy:
        best_accuracy = accuracy
        best_model = pipeline
    
print('Best model: ',best_model)
```
