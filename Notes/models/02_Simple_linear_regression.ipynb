#linear regression

`Linear Regression` is a supervised machine learning where the predicted output is `continuous` and has a `constant slope`. Its used to predict values with in a continuous range, (e.g sales price ) rather than trying to classify them into categories (e.g dog, cat). There are two main types:

1- Simple linear regression <br/>
2- Multi linear regression

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score,mean_squared_error
```

```python
# load the data
df = sns.load_dataset('tips')
df.head()
```

```python
sns.scatterplot(x='total_bill', y='tip', data=df)
```

```python
# split the data into x and y
x = df[['total_bill']]
y = df['tip']
```

```python
# split the data into train and test
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2)
```

```python
# call the mode
model = LinearRegression()
```

```python
# train the model
model.fit(x_train,y_train)
```

```python
# take out model slope and intercept
print(model.coef_)
print(model.intercept_)
print(f"y= ",model.intercept_," + ", model.coef_ , "*x")
```

```python
# predict
y_pred = model.predict(x_test)
```

```python
# evalute the error
print('MSE: ', mean_squared_error(y_test,y_pred))
print('RMSE: ', np.sqrt(mean_squared_error(y_test,y_pred)))
print('R2: ', r2_score(y_test,y_pred))
```

```python
# plot the model
plt.scatter(x_test,y_test)
plt.plot(x_test,y_pred,color='red')
plt.show()
```
