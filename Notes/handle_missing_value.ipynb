# Handling Missing Values
There are numerous techniques for handling missing data. We will focus on the most commonly used methods:

1. **Mean, Median, and Mode Imputation**
2. **Scikit-Learn Imputers (SimpleImputer, KNNImputer)**
3. **Dropping Rows or Columns**
4. **Forward Fill (ffill) and Backward Fill (bfill)**

### 1. Mean, Median, and Mode Imputation
This technique involves replacing missing values with a calculated statistic from the column.

- **Mean**: Used for numerical data with a normal distribution (no extreme outliers).
- **Median**: Best for numerical data with outliers, as it is robust to extreme values.
- **Mode**: Ideal for categorical (non-numerical) data.

```python
import numpy as np
import pandas as pd
import seaborn as sns

# Load the dataset
data = sns.load_dataset('titanic')

# Check for missing values
print(data.isnull().sum().sort_values(ascending=False))
```

```python
# filling with median
data['age']= data['age'].fillna(data['age'].median())
print(data.isnull().sum().sort_values(ascending=False))
```

```python
# filling embarked town with mode
data['embark_town'] = data['embark_town'].fillna(data['embark_town'].mode()[0])

print(data.isnull().sum().sort_values(ascending=False))
```

### 2. Handling Missing Values using Scikit-Learn (`sklearn`)
We can also use the `sklearn.impute` module for more advanced imputation strategies:
- **Univariate Imputation (`SimpleImputer`)**: Imputes missing values using a single column's statistics (e.g., mean, median).
- **Multivariate Imputation (`KNNImputer`)**: Imputes missing values by analyzing the K-Nearest Neighbors across all features in a row.

```python
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.impute import SimpleImputer
from sklearn.impute import KNNImputer

data = sns.load_dataset('titanic')

print(data.isnull().sum().sort_values(ascending=False))
```

```python

imputer = SimpleImputer(strategy='median')

data['age'] = imputer.fit_transform(data[['age']]).ravel()
print(data.isnull().sum().sort_values(ascending=False))
```

```python
imputer = KNNImputer()

data[['age']] = imputer.fit_transform(data[['age']])
print(data.isnull().sum().sort_values(ascending=False))
```

### 3. Dropping Rows or Columns
If a column or row contains too many missing values, or if the missing data cannot be reliably imputed, removing the missing data using `dropna()` is often the best approach.

```python
import numpy as np
import pandas as pd
import seaborn as sns

data = sns.load_dataset('titanic')
print(data.isnull().sum().sort_values(ascending=False))
```

```python
data.dropna(subset = 'age',inplace=True)
print(data.isnull().sum().sort_values(ascending=False))
```

### 4. Forward Fill (ffill) and Backward Fill (bfill)
Often used in sequential or time-series data:
- **Forward Fill (`ffill`)**: Propagates the last valid observation forward to fill gaps.
- **Backward Fill (`bfill`)**: Uses the next valid observation to fill gaps backward.

```python
import numpy as np
import pandas as pd
import seaborn as sns

data = sns.load_dataset('titanic')

```

```python
data['age'] = data['age'].ffill()
print(data.isnull().sum().sort_values(ascending=False))
```

```python

data['age'] = data['age'].bfill()
print(data.isnull().sum().sort_values(ascending=False))
```
