# Normalization
Normalization is a technique used to adjust the values in a dataset to a common scale. Unlike feature scaling (which is usually column-wise), the **Normalizer** in Scikit-Learn works **sample-wise** (row-wise), ensuring each row has a unit norm.

### 1. L2 Normalization (Euclidean Norm)
L2 normalization rescales each row so that the sum of the squares of its elements equals 1. This is commonly used in text classification and clustering to ensure that documents of different lengths are comparable.

- **Formula**: $x_{norm} = \frac{x}{\sqrt{\sum x_i^2}}$
- **Library**: `sklearn.preprocessing.Normalizer(norm='l2')`

```python
from sklearn.preprocessing import Normalizer
data = [[1,1,1],[1,1,0],[1,0,0]]

normalizer = Normalizer(norm='l2')
print(normalizer.fit_transform(data))
```

### 2. L1 Normalization (Manhattan Norm)
L1 normalization rescales each row so that the sum of the absolute values of its elements equals 1. It is useful when the relative importance of features is linear rather than squared.

- **Formula**: $x_{norm} = \frac{x}{\sum |x_i|}$
- **Library**: `sklearn.preprocessing.Normalizer(norm='l1')`

```python
from sklearn.preprocessing import Normalizer
data = [[1,1,1],[1,1,0],[1,0,0]]

normalizer = Normalizer(norm='l1')
print(normalizer.fit_transform(data))
```

### 3. Scaling vs. Normalization
While "Normalization" in `sklearn` refers to row-wise scaling, in broader contexts it often includes common scaling techniques:

- **Standardization (Z-Score)**: Centering data around mean 0 with unit variance.
- **Min-Max Scaling**: Scaling data to a fixed range, typically [0, 1].

### 4. Log Transformation
Log transformation is used to handle skewed data and compress the range of values. It helps in stabilizing variance and making the data distribution more symmetrical (normal-like).

- **Formula**: $y = \log(x)$

```python
from os import name
import pandas as pd
import numpy as np
import seaborn as sns

# example with skewed values
df = {"values": [1,5,10,20,50,100,1000,2000,5000,10000,20000]}

df = pd.DataFrame(df)
sns.histplot(df["values"],kde=True)
```

```python
df['log_values'] = np.log(df['values'])
sns.histplot(df['log_values'],kde=True)
df
```
