# Data Integration
Data integration is the process of combining data from multiple sources into a single, unified view. It is a critical step in data preprocessing, ensuring that all necessary information is available for analysis and modeling.

We will focus on the most commonly used methods:
1. **Data Merging**
2. **Data Concatenation**

### Why Data Integration?
- **Holistic View**: Provides a complete picture by combining related data from different sources.
- **Improved Accuracy**: Helps cross-verify data and fill in gaps.
- **Efficiency**: Centralizing data makes it easier to process and analyze.

### Challenges in Data Integration
- **Entity Identification**: Matching the same real-world entity across different formats (e.g., "Alice" vs "Alice Smith").
- **Redundancy**: Identifying and removing duplicate records or overlapping attributes.
- **Data Value Conflicts**: Resolving cases where different sources provide conflicting information for the same record.

````python
import pandas as pd
import numpy as np
df1 = {
    'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'age': [25, 30, 35, 40, 45]
}
df2 = {
    'city': ['New York', 'San Francisco', 'Los Angeles', 'Chicago', 'Houston'],
    'salary': [50000, 60000, 70000, 80000, 90000]
}

df1 = pd.DataFrame(df1)
df2 = pd.DataFrame(df2)

merged_data =pd.merge(df1,df2, on="id", how="inner")
merged_data
````

### 1. Data Merging
Merging is used to combine datasets based on a common key or index. It is similar to the `JOIN` operation in SQL. This technique is ideal when you have related data in different tables that you want to bring together.

**Common Types of Joins:**
- **Inner Join**: Keeps only the rows where the key exists in both datasets.
- **Outer Join**: Keeps all rows from both datasets, filling missing values with `NaN`.
- **Left Join**: Keeps all rows from the left dataset and matching rows from the right.
- **Right Join**: Keeps all rows from the right dataset and matching rows from the left.

### 2. Data Concatenation
Concatenation involves "stacking" datasets either vertically (adding rows) or horizontally (adding columns). It is most effective when datasets share the same structure or features (e.g., merging monthly sales reports into a yearly summary).

**Basic Usage:**
```python
# Vertically stacking data
result = pd.concat([df1, df2], axis=0)
```
